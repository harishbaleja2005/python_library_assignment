```python
#4.Download a large Excel file (at least 10,000 rows) of Flipkart product listings. Use pd.read_excel() with
# the chunksize parameter to read the file in chunks of 2000 rows, and print the number of rows in each chunk as you iterate.
import pandas as pd

data=pd.read_excel('flipkart.xlsx')

for i in range(0, len(data), 2000):
    chunk = data.iloc[i:i+2000]
    print("number of rows:", len(chunk))
```
