```python
#1.Given a list of delivery timestamps as strings (e.g., ['2024-06-01 14:30', '2024-06-02 09:15', '2024-06-03 20:45']),
# use pandas and pd.to_datetime() to convert them into datetime objects and print the result.
import pandas as pd

data = pd.DataFrame({
    "delivery_time": [
        "2024-06-01 14:30",
        "2024-06-02 09:15",
        "2024-06-03 20:45"
    ]
})

data["delivery_time"] = pd.to_datetime(data["delivery_time"])

print(data)
```

```python
#2.Load a CSV file containing order dates from a Flipkart-style order history (column: 'order_date', format: 'YYYY-MM-DD HH:MM:SS').
# Extract the year, month, and weekday for each order and add them as new columns in the DataFrame.
import pandas as pd

data = pd.read_csv("flipkart_order.csv")

data["order_date"] = pd.to_datetime(data["order_date"])

data["year"] = data["order_date"].dt.year
data["month"] = data["order_date"].dt.month
data["weekday"] = data["order_date"].dt.day_name()

print(data)
```

```python
#3.Set the 'order_date' column as the index of your DataFrame and use resampling to calculate the total number
# of orders placed each week.<br><br><em><strong>Hint:</strong> Use df.resample('W').size() after setting the datetime index.</em>
import pandas as pd

data = pd.read_csv("flipkart_order.csv")

data["order_date"] = pd.to_datetime(data["order_date"])

data = data.set_index("order_date")

weekly_orders = data.resample("W").size()

print(weekly_orders)
```

```python
#4.Suppose you have a DataFrame of Instagram posts with a 'posted_at' column in UTC. Convert these timestamps
# to 'Asia/Kolkata' timezone and display the first 5 converted times.
import pandas as pd

data = pd.DataFrame({
    "posted_at": [
        "2024-06-01 14:30:00",
        "2024-06-02 09:15:00",
        "2024-06-03 20:45:00",
        "2024-06-04 11:00:00",
        "2024-06-05 16:30:00",
        "2024-06-06 08:45:00"
    ]
})

data["posted_at"] = pd.to_datetime(data["posted_at"], utc=True)

data["posted_at"] = data["posted_at"].dt.tz_convert("Asia/Kolkata")

print(data["posted_at"].head(5))
```

```python
#5.Create a new feature called 'is_weekend' in your DataFrame that marks True if an order was placed on Saturday
# or Sunday, and False otherwise.<br><br><em><strong>Constraint:</strong> Do not use any external libraries except pandas and numpy.</em>
import pandas as pd

data = pd.DataFrame({
    "order_date": [
        "2024-06-01",
        "2024-06-03",
        "2024-06-08",
        "2024-06-09",
        "2024-06-10"
    ]
})

data["order_date"] = pd.to_datetime(data["order_date"])

data["is_weekend"] = data["order_date"].dt.dayofweek >= 5

print(data)
```
