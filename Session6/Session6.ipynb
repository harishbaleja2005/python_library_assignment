```python
#1.Download a sample CSV file of IPL player stats (include columns like player name, runs, matches, and some
# missing values). Load it into a pandas DataFrame and use isnull() and notnull() to print how many missing values are present in each column.
import pandas as pd

data=pd.read_csv('ipl_player_stats.csv')
print(data.isnull().sum())
print(data.notnull().sum())
data.head()
```

```python
#2.Using the loaded IPL player stats DataFrame, apply dropna(axis=0, how='any') to remove all rows with any
# missing data and display the shape of the DataFrame before and after dropping.
import pandas as pd

data=pd.read_csv('ipl_player_stats.csv')
print("shape before dropping:",data.shape)
df=data.dropna(axis=0,how="any")
print("shape after dropping:",df.shape)
print(df.head())
```

```python
#3.For the 'runs' column in your IPL player stats DataFrame, use fillna() to replace missing values with the 
# mean of the column. Print the updated column to verify the changes.
import pandas as pd

data=pd.read_csv('ipl_player_stats.csv')
data["runs"]=data["runs"].fillna(data["runs"].mean())
print("after fill null values:\n",data["runs"])
```

```python
#4.Simulate a Zomato-style restaurant ratings dataset with some missing ratings. Use forward fill (method='ffill')
# to fill missing values in the ratings column, then use backward fill (method='bfill') for any remaining missing values. Show the before and after results.
import pandas as pd

data=pd.DataFrame({
    "restaurant": ["Domino's", "McDonald's", "Subway", "Haldiram's", "Barbeque Nation", "La Pino's"],
    "ratings": [4.5, None, 4.0, None, None, 3.5]
})

print("before filling data:")
print(data)

data["ratings"]=data["ratings"].ffill()
data["ratings"]=data["ratings"].bfill()

print("after filling:")
print(data)
```
